```python
#Hypothesis: I predict that 90% of the positions in this dataframe are with 3 km of Olin Hall

#Import plotly and math to do the map circle illustration
import plotly.express as px
import plotly.graph_objects as go
import math

#Save Olin Hall's coordinates
olin_hall_lat = 46.0728341
olin_hall_lon = -118.3290384
olin_hall_coords = (olin_hall_lat, olin_hall_lon)

# Radius for the circle (3 km)
radius_km = 3

#Write a function to take the coordinates and desired radius of the circle
def get_circle_points(lat, lon, radius_km, num_points=200):
    """Generates points for a circle of a given radius (in km) around a center (lat, lon)."""
    points = []
    # Earth's radius in kilometers
    R = 6371.0
    # Convert radius and center coordinates to radians
    lat_rad = math.radians(lat)
    lon_rad = math.radians(lon)

    #Iterate over the range of points to get their approx position in lat. long format
    for i in range(num_points):
        bearing = math.radians(i * (360 / num_points))  # Bearing from 0 to 360 degrees

        # Calculate new latitude
        new_lat_rad = math.asin(math.sin(lat_rad) * math.cos(radius_km / R) +
                                math.cos(lat_rad) * math.sin(radius_km / R) * math.cos(bearing))

        # Calculate new longitude
        new_lon_rad = lon_rad + math.atan2(math.sin(bearing) * math.sin(radius_km / R) * math.cos(lat_rad),
                                        math.cos(radius_km / R) - math.sin(lat_rad) * math.sin(new_lat_rad))

        points.append((math.degrees(new_lat_rad), math.degrees(new_lon_rad)))
    return points

# Build the points for the 30km circle around Olin Hall
circle_points = get_circle_points(olin_hall_lat, olin_hall_lon, radius_km)
circle_lats, circle_lons = zip(*circle_points)

# Create the Plotly map
fig = px.scatter_mapbox(df,
                        lat="lat",
                        lon="lon",
                        zoom=8,
                        mapbox_style="open-street-map",
                        title=f'Prof WB Location Data with {radius_km}km Radius around Olin Hall')

# Add the 30km radius circle around Olin Hall
fig.add_trace(go.Scattermapbox(
    lat=circle_lats,
    lon=circle_lons,
    mode="lines",
    line=dict(width=2, color="red"),
    name=f'{radius_km}km Radius around Olin Hall',
    hoverinfo='name'
))

# Add a marker for Olin Hall
fig.add_trace(go.Scattermapbox(
    lat=[olin_hall_lat],
    lon=[olin_hall_lon],
    mode="markers",
    marker=go.scattermapbox.Marker(size=10, color='green', symbol='star'),
    name="Olin Hall",
    hoverinfo='name'
))

fig.show()
```
